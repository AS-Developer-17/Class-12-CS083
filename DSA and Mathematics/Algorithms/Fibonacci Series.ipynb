# Generating a Fibonacci

> This program is method for creating series by For Loop

```python
valA = 0
valB = 1
series= []

for fibCnt in range(1,10):
    seriesVal = valA + valB
    valA = valB
    valB = seriesVal
    series.append(seriesVal)
print(series)
```

> This program is for creating series by Recursion

```python
series=[]
countVal = 2
def fibGen(valA, valB):
    global countVal,series
    if countVal <= 10: 
        seriesVal = valA+valB
        series.append(seriesVal)
        valB= valA 
        valA= seriesVal
        countVal+=1
        fibGen(valA,valB)
        print(series)
    else:
        print(series)

fibGen(1,0)
```
