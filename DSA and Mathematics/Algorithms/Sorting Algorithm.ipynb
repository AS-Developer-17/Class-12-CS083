# *Sorting Algorithms*

> Sorting Algorithms are useful for sort the array content in a specific order.

'This file contains most of the useful algorithms for sorting'

```python
# Array Generator Function
''' This function allows us to generate a array for sorting purposes'''
from random import randint
def arrayGen():
    arrayVal =[]
    for arrayGenCnt in range(1, 1000):
        arrayVal.append(randint(0,randint(100,1000)))
    return arrayVal
```

## Bubble Sort

>This Algorithm works by swapping the consequtive larger and smaller values.

>It uses iterative approach to do so.

```python
# Bubble Sort Code

bsArray = arrayGen()

print(f"The orignal array is : {bsArray}")
count = 0
while True:
    for travVal in range(0,len(bsArray)-1):
        
        if bsArray[travVal]>bsArray[travVal+1]:
            # Swapping Functions
            val = bsArray[travVal]
            bsArray[travVal]= bsArray[travVal+1]
            bsArray[travVal+1]=val
            count += 1
        
    print(bsArray[travVal],bsArray[travVal+1])
    print(bsArray, count)
    if count==0 :
        break
    else:
        continue
    count=0 
print(f"The sorted array in ascending order is : {bsArray}")
```

# Selection Sort

>This algorithm starts by checking the smalest value then <br> performs operation of swapping or shifting to sort the array.

```python
# Selection Sort

ssArray = arrayGen()

while True:
    shifting=False
    minVal = ssArray[0]
    iMin= minVal
    for travVal in range(0,len(ssArray)-1):
        if minVal > ssArray[travVal]:
            minVal= ssArray[travVal]
            indexVal= travVal
            print("The minimum value in the array is ",minVal," at the index of ",indexVal," .")
    
    critVal = ssArray[indexVal]
    for shift in range(1,indexVal):
        ssArray[shift]= ssArray[shift-1]
    
    if iMin == minVal:
        shifting =True

    ssArray[0]=critVal
    print(f'{ssArray}')
    
    if shifting== True:
        continue
    else:
        break
```

# Insertion Sort

> This algorithm works by creating a sorted part and a unsorted one which is managed later.

> Major operation of this algorithm is sorting

```python
# Insertion Sort

iArray= arrayGen()
print("The orignal array is: ",iArray)
for me in range(0,len(iArray)):
    indexVal=int()

    for newMe in range(0,me):
        if iArray[me] < iArray[newMe]:
            indexVal= newMe
            break
        
    cVal = iArray[me]
    for newCnt in range(indexVal+1 , me):
        iArray[newCnt]= iArray[newCnt-1]
    iArray[indexVal]=cVal

    print(iArray)                
    
    
print(f"The sorted array is: {iArray}")
```
