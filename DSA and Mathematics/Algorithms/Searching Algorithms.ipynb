# **Linear Search**

```python
from random import randint

def arrayGen():
    arrayVal =[]
    for me in range(0,99):
        arrayVal.append(randint(1,randint(2,10)))
    return arrayVal

arr= arrayGen()

def linearSearch(searchVal,array):
    indexList=list()
    
    for searching in range(0,len(array)-1):
        if searchVal == array[searching]:
            indexList.append(searching)
        else:
            continue
    if len(array)==0 :
        return "Couldn't find that value in that array."
    else:
        return indexList

print("The orignal array is \n", arr)    
searchVal=int(input("Type the value you want to search"))
linearSearch(searchVal, arr)
```
