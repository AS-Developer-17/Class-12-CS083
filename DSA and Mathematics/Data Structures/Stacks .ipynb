# Stacks (Data Structure)

> They are analogous to the stack of books, pancakes in real life cases. 

>They allow us to perform operations like : 


|Operation|Use Case|
|---|----|
|Push| Adds a new element on the stack.|
|Pop| Removes and returns the top element from the stack.|
|Peek| Returns the top element on the stack.|
|isEmpty| Checks if the stack is empty.|
|Size| Finds the number of elements in the stack.|



```python
from random import randint 
rndVal= randint(0,1000)
```

```python
# Basic Implementation of Stacks

array = list()

# Push Operation
print(array)
for me in range(0,5):
    rndVal= randint(0,1000)
    array.append(rndVal)

print(f"Peek Operator: {array[-1]}")

print(array)
for me in range(0,5):
    array.pop()

print(array)

lenVal= len(array)

print(f"The size of array is {lenVal}")

if lenVal ==0 :
    print("Empty Array")
else: 
    print("Array is not empty")
```
