# Arrays (Data Structure)

### Uses of Array
> List or Tuples allowing us to store multiple elements. 

> Frequently use by many operations like (sort, deque, enque etc) and various algorithms.



```python
from random import randint

def rndArrayGen():
    rndArray= []
    for arrayGen in range(0,9):
        rndArray.append(randint(0,100))
    return rndArray
```

```python
# Minimum Value Finder in Array

arrayVal = rndArrayGen()
print("The array is : ",arrayVal)

minVal= arrayVal[0]
for trav in arrayVal:
    if minVal >= trav:
        minVal= trav 
    print(f"The minimum value in the list is {minVal}. ")
```

```python
arrayVal= rndArrayGen()
def maxValFinder(arr):
    maxVal = arr[0]
    for trav in arr:
        if maxVal < trav:
            maxVal=trav
    return maxVal
print("The orignal array is: ",arrayVal)
print(f"The max value is {maxValFinder(arrayVal)}")
```
