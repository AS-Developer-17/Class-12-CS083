# **Trees**

> This Data Structure is similar to Linked List in layman's term.

### Uses of Trees

|Field of Use| Use Cases |
|-----|------|
|Hierarchical Data| File systems, organizational models, etc.|
|Databases| Used for quick data retrieval.|
|Routing Tables| Used for routing data in network algorithms.|
|Sorting/Searching| Used for sorting data and searching for data. |
|Priority Queues| Priority queue data structures are commonly implemented using trees, such as binary heaps.|

### Facts Associated with Trees

> First Node == Root Node

> Connection among Nodes is achieved by **Edges**.

>A parent or internal node has links to its child nodes. 

>A node can have zero, one, or many child nodes.

>A node can only have one parent node.

> Childs(Node)== 1 :: Leaf Node  

> Height(Tree) = Max(Number(Edges)) from Root to Leaf Node 

> Size(Tree) =  number of nodes 

### Types of Tree

|Name|Description|
|----|-----------|
|Binary Trees| Max(Child(Node))==2 ; Right and Left Child |
|Binary Search Trees (BSTs)|Val(Node(Left)) < Val(Node(Right)) |
|AVL Trees| A balanced BST where Height(Left Tree) - Height(Right Tree) = 1 |

### Type of Binary Trees
> A Balanced BT | Height(Right SubTree) - Height(Left SubTree) | <= 1.

> Binary Tree have Child(Node) == 0 or 2 child.



```python
# Binary Tree Implementation

class TreesInPython():
    def __init__(self,data):
        self.data= data
        self.left= None
        self.right= None 
root = TreesInPython('R')

NodeA=TreesInPython("A")
NodeB=TreesInPython("B")
NodeC=TreesInPython("C")
NodeD=TreesInPython("D")
NodeE=TreesInPython("E")
NodeF=TreesInPython("F")
NodeG=TreesInPython("G")

root.left = NodeA
root.right= NodeB

NodeA.right= NodeD
NodeA.left= NodeC

NodeB.right = NodeF 
NodeB.left  = NodeE
NodeF.right = NodeG

```

# Ways to Traverse Binary Tree

There are two main categories of Tree traversal methods:

> **Breadth First Search (BFS)** : is when the nodes on the same level are visited before going to the next level in the tree. This means that the tree is explored in a more sideways direction.

> **Depth First Search (DFS)** : is when the traversal moves down the tree all the way to the leaf nodes, exploring the tree branch by branch in a downwards direction.

### Different Type of DFS traversals:

> pre-order

> in-order

> post-order

## Pre-Order-Traversal
Visit the root node first then recursively do pre-order traversal of left subtree, followed by traversal of right one.




```python
def preOrderTraversal(node):
    if node is None:
        return 
    print(node.data,end=" ")
    preOrderTraversal(node.left)
    preOrderTraversal(node.right)
    
preOrderTraversal(root)
```

# In-Order-Traversal

In-order Traversal does a recursive In-order Traversal of the left subtree, visits the root node, and finally, does a recursive In-order Traversal of the right subtree. This traversal is mainly used for Binary Search Trees where it returns values in ascending order.

What makes this traversal "in" order, is that the node is visited in between the recursive function calls. The node is visited after the In-order Traversal of the left subtree, and before the In-order Traversal of the right subtree.

```python
def inOrderTraversal(node):
    if node is None:
        return
    inOrderTraversal(node.left)
    print(node.data, end=", ")
    inOrderTraversal(node.right)
    
inOrderTraversal(root)
```

# Post-order Traversal

Post-order Traversal works by recursively doing a Post-order Traversal of the left subtree and the right subtree, followed by a visit to the root node. It is used for deleting a tree, post-fix notation of an expression tree, etc.

What makes this traversal "post" is that visiting a node is done "after" the left and right child nodes are called recursively.

```python
def postOrderTraversal(node):
    if node is None:
        return
    postOrderTraversal(node.left)
    postOrderTraversal(node.right)
    print(node.data, end=", ")
postOrderTraversal(root)
```
