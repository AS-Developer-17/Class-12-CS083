# Terminology and Basic Idea

### Figure Related
---
> **Figure** is area where graph is ploted, which have axis as per need later when the graph is ploted it is required to be shown by the plot.show function.

> **Title** : It is set by the .set_title

> **Label** : They are the axis specified labels.

> **Spines**: They allow us to show or hide the boundary of the plot

> **TickLabel**: For measuring tiny bits on our axis

> **Legend** : Allows us to gather more information about the plot.
---

### Plot Types

|Type|Function|Brief Description|
|---|----|----|
|Lines|Plot()|Allowing us to generate curve or plot a function.|
|Markers|Scatter()|Alllows us to plot a scatter plot for a given data. |
|Axes|Subplots()|Artist attatched to figure with region of plotting data.|
