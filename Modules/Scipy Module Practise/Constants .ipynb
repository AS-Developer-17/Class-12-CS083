```python
import scipy
import scipy.constants

# Mathematical Constants
print("The value of 'e' is : ",scipy.constants.e)
print("The value of 'pi' is : ",scipy.constants.pi)
print("The value of 'golden ratio' is : ",scipy.constants.golden)

```

```python
# Physic's Constant
print("The value of 'c' is : ",scipy.constants.c)
print("The value of 'mu_0' is : ",scipy.constants.mu_0)
print("The value of 'epsilon_0' is : ",scipy.constants.epsilon_0)
print("The value of 'h bar' is : ",scipy.constants.hbar)
print("The value of 'G' is : ",scipy.constants.G)
print("The value of 'e' is : ",scipy.constants.elementary_charge)
```

```python
# Chemical Constant
print("The value of 'Avagadro Number' is : ",scipy.constants.N_A)
print("The value of 'R' is : ",scipy.constants.R)
print("The value of 'Rydberg' is : ",scipy.constants.Rydberg)
print("The value of 'Neutron Mass' is : ",scipy.constants.m_n)
print("The value of 'Electron Mass' is : ",scipy.constants.m_e)
print("The value of 'K' is : ",scipy.constants.k)
```

# **Scipy.Constants**
***
## Methods
> *Values* : Help us to get constant's value.

> *Find*   : ALlows to access the related search terms.

> *Precision* : Allows us to get the precision value for different values.

> *Unit*   : Generates the unit for the search term.

---

## Conversion Functions

>*nu2lambda* : Help us to reedeem Wavelength from wave number. 

>*lambda2nu* : Help us to reedeem wave number from Wavelength.

>*convert_temperature* : Allows us to switch between different unit of temperature measurement.

---

```python
# Scipy.Constants.Value

scipy.constants.value('electron mass')
```

```python
# Scipy.Constants.Unit 

scipy.constants.unit("proton mass")
```

```python
# Scipy.Constants.Precision 

scipy.constants.precision("proton mass")
```

```python
# Scipy.Constants.Find 

scipy.constants.find("radian")
```

```python
# Scipy.Constants.ConstantWarning

"""
Occurs due to the fact that 
variable being acessed is no 
longer part of the CODATA data 
set. (Means Not Availiable)
"""
```

```python
# Conversion Operators

# Temperature Conversions
scipy.constants.convert_temperature(16,'Kelvin','celsius')
'''
Key Value Pair for Temperature Units

Celsius :    C
Fahrenheit : F
Rankine :    R  
Kelvin  :    K
'''

```

```python
# Wavelength to Frequency (Vice Versa)

#  Frequency to Wavelength Conversion
scipy.constants.nu2lambda((111))

#  Wavelength to Frequency Conversion
scipy.constants.lambda2nu((111))
```
