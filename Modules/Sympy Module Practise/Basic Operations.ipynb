```python
from sympy import *
import numpy
```

```python
# Variable Decleration
x,y,z = symbols("x y z")

# Substitution
expr = cos(x+y)+11
print(f"The substituted value of the expression is: {expr.subs(x,y)}")  

# Trignometeric Substitution
exprTrig = cos(x*2) + cos(x*2)
print(f"Substitution: {expr.subs(x,expand_trig(exprTrig))}")
 
# Multiple Substitution
exprMV=x**3+((y*x)/z) + (x**z) - z**3-z/y
subEx = exprMV.subs([(z,12),(x,100),(y,1234)])
print(subEx)
 
```

```python
# Strings to Expression Conversion
str_expr= "x**2+3*x-1/2"
expr= sympify(str_expr)
print(expr)

# Evalf for Evaluation of Numerical Expression

expr = sqrt(7)
print(expr.evalf(50),pi.evalf(50))
```
