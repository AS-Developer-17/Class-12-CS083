```python
import matplotlib.pyplot as plt
import numpy
from sympy import *

# Variable Decleration
x,y,z = symbols("x y z")
```

```python
a=simplify(sin(x)**2+cos(x)**2)
b=simplify((x**3+x**2-x+1)/(x**2+x*2+1))
c=simplify(gamma(z)/gamma(z-2))
print(a,b,c, sep="\n")
```

```python
# Expansion 
expand((z-2*x-y*x*z+z)**2)
expand((x+2)*(x-2))

# Factorization
factor(x**3-x**2-x+12)
factor(x*y-x*z**2-3*z*x*y)

# Collect
expr= x*y +x-3+2*x**2-z*x**2+x**3
print(expr)
collectedExpr= collect(expr,x)
print(collectedExpr)

#Coefficient Method
print(collectedExpr.coeff(x,2) )

# Cancel Method
'''This method is use to get the simplest rational form'''
a= expand(((x+2)**2)*(x+3))
b= expand(((x+3)**2)*(x+2))
print(cancel(a/b))

# Apart Method
'''Performs the partial decomposition of rational function'''
expr= (4*x**3+21*x**2+10*x+12)/(x**4+5*x**3+5*x**2+4*x)
print(f"Normal Expression : {expr}")
print(f"Normal Expression : {apart(expr)}")

# Trignometeric Simplification
acos(x) # Inverse Cosine function

trigsimp(sin(x)**2+cos(x)**2) # Simplification of Trignometeric Expression 

expand_trig(sin(x+z)) # Expansion of Trignometeric Identities
```
