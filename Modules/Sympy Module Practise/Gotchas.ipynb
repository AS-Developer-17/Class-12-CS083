```python
# Variable Decleration 
# Unlike many symbolic manipulation systems you may have used, in SymPy, variables are not defined automatically.
x= sympy.symbols("x")

# symbols takes a string of variable names separated by spaces or commas, and creates Symbols out of them.
nu,y,z= sympy.symbols("nu y z")
```

```python
# Expression Declerations

expr = x**2 + 7*x + 10 
print(f"Expression (Normal):  {expr}")

# Evaluating Function

'''By Substitution'''
exprSubs= expr.subs(x,1)
print(f"Expression (Substituted): {exprSubs}")

'''By Alloting Values Nothing Changes''' 
x= 12 
print(f"Expression (Non Substituted): {expr}")
```

```python
# Equal Sign 
x= 14
equality= x+1 == 15
print(equality)
equality= x+1 == 14
print(equality)

# Sympy Equality
print(sympy.Eq(x+1,15))

'''
(x + 1)**2 == x**2 + 2*x + 1
For expression like these we need to 
simplify the expression and check
'''
a= (x+1)**2 
b= x**2 + 2*x + 1
print(sympy.Eq(sympy.simplify(a-b),0))

a= sympy.cos(x)**2 -sympy.sin(x)**2
b= sympy.cos(2*x)
print(a.equals(b))
```
